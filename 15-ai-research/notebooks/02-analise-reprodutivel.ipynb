# Análise Reprodutível com Dataset Sintético
Objetivo: mostrar que **código que executa não basta**. Precisamos documentar dados, decisões, seed e método.


```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
SEED = 20260907
rng = np.random.default_rng(SEED)
path = Path('../dados/dataset-sintetico.csv')
df = pd.read_csv(path)
df.head()
```

```python
df.info()
df.groupby('grupo')['delta'].agg(['count','mean','std','median'])
```

## Visualização


```python
for g, sub in df.groupby('grupo'):
    plt.figure()
    plt.hist(sub['delta'], bins=12, alpha=0.8)
    plt.title(f'Distribuição do delta — {g}')
    plt.xlabel('delta')
    plt.ylabel('frequência')
    plt.show()
```

## Diferença de médias e bootstrap
O bootstrap abaixo é um exemplo didático. Antes de aplicá-lo a dados reais, discuta desenho, independência e estimando de interesse.


```python
a = df.loc[df.grupo=='intervencao','delta'].to_numpy()
b = df.loc[df.grupo=='controle','delta'].to_numpy()
diff = a.mean() - b.mean()
B = 5000
boots = np.empty(B)
for i in range(B):
    aa = rng.choice(a, size=len(a), replace=True)
    bb = rng.choice(b, size=len(b), replace=True)
    boots[i] = aa.mean() - bb.mean()
ci = np.quantile(boots, [0.025, 0.975])
print('diferença de médias:', round(diff,3))
print('IC bootstrap 95%:', np.round(ci,3))
```
