Este notebook não confirma referências pela internet. Ele prepara uma planilha de auditoria e ensina a separar **metadados**, **existência** e **suporte da afirmação**.


```python
import pandas as pd
refs = pd.DataFrame([
    {'titulo':'[cole aqui]','autores':'','ano':'','periodico':'','doi':'','afirmacao_associada':''},
])
refs
```

## Campos de verificação
A confirmação real deve ser feita em fonte bibliográfica/DOI e, para suporte científico, no artigo original.


```python
for col in ['titulo_confere','autores_conferem','doi_resolve','fonte_original_localizada','afirmacao_suportada','decisao','observacoes']:
    refs[col] = ''
refs
```

## Validação de formato de DOI (não de existência)


```python
import re
doi_pattern = re.compile(r'^10\.\d{4,9}/\S+$', re.I)
refs['doi_formato_plausivel'] = refs['doi'].fillna('').apply(lambda x: bool(doi_pattern.match(str(x).strip())))
refs[['doi','doi_formato_plausivel']]
```
